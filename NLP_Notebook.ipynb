```python
#Create Sentiment Labels from Ratings
def label_sentiment(rating):
    if rating >= 4:
        return "Positive"
    elif rating == 3:
        return "Neutral"
    else:
        return "Negative"
```

```python
df.head()
```

```python
df.drop(columns=['helpful_votes', 'platform', 'version', 'verified_purchase'])
```

```python
df=df.drop(columns=['helpful_votes', 'platform', 'version', 'verified_purchase'])
```

```python
df.columns
```

```python
df['sentiment'] = df['rating'].apply(label_sentiment)
```

```python
X=df['review']
y=df['sentiment']
```

```python
cv = CountVectorizer()
X = cv.fit_transform(X) # Fit the Data
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y)
```

```python
clf = MultinomialNB()
clf.fit(X_train,y_train)
```

```python
data = cv.transform(['Wait address exactly push film. Prevent growth appear tell western PM. Trial unit concern before.']).toarray()
a = clf.predict(data)
print(a)
```
